```python
def clean_text(text):
    if not isinstance(text, str): #if text is not string(dataset may contain NaN, None values), then it will return an empty string
        return ""
    text = text.lower()
    #we will have to remove words like remastered, delux, live, etc.
    text = re.sub(r"\(.*?\)", "", text) #will remove everythng inside () brackets
    text = re.sub(r"\[.*?\]", "", text) #will remove everything inside [] brackets
    text = re.sub(r"[^a-z0-9\s]", "", text) #will remove everything except a-z and 0-9, used for removing punctuations, emojis, etc.
    text = re.sub(r"\s+", " ", text) #replaces extra spaces, tabs and new line with single white space
    return text.strip()
```

```python
df_global = spotify_global.copy()

df_global = df_global.drop(columns=["Unnamed: 0"], errors="ignore")
df_global = df_global.rename(columns={"artists": "artist_name"})

df_global = df_global.dropna(subset=[
    "track_name",
    "artist_name",
    "track_genre"
]) #drop rows where anyone of the above is missing

df_global["track_clean"] = df_global["track_name"].apply(clean_text)
df_global["artist_clean"] = df_global["artist_name"].apply(clean_text)

df_global["canonical_key"] = (
    df_global["track_clean"] + "_" + df_global["artist_clean"]
)

df_global["dataset_source"] = "global"

df_global = df_global.reset_index(drop=True) #removes old index and reassigns new 0-based index
```

```python
df_regional = spotify_regional.copy()

df_regional = df_regional.dropna(subset=[
    "track_name",
    "artist_name"
])

df_regional["track_clean"] = df_regional["track_name"].apply(clean_text)
df_regional["artist_clean"] = df_regional["artist_name"].apply(clean_text)

df_regional["canonical_key"] = (
    df_regional["track_clean"] + "_" + df_regional["artist_clean"]
)

df_regional["dataset_source"] = "regional"

df_regional = df_regional.reset_index(drop=True)
```

```python
common_columns = [
    "track_id",
    "track_name",
    "artist_name",
    "album_name",
    "popularity",
    "duration_ms",
    "explicit",
    "danceability",
    "energy",
    "key",
    "loudness",
    "mode",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
    "time_signature",
    "track_genre",      # global only
    "language",         # regional only
    "track_url",        # regional only
    "artwork_url",      # regional only
    "canonical_key",
    "dataset_source"
]
```

```python
for col in common_columns:
    #if a column exist in df_regional, but is missing in df_global, then you add col in df_global and fill it with NaN
    if col not in df_global.columns:
        df_global[col] = np.nan
    #if a column exist in df_global, but is missing in df_regional, then you add col in df_regional and fill it with NaN
    if col not in df_regional.columns:
        df_regional[col] = np.nan
```

```python
df_combined = pd.concat(
    [df_global[common_columns], df_regional[common_columns]],
    ignore_index=True #ignores old indices and creates new ones
)
#this concatenates both datasets row wise
```

```python
before = len(df_combined)
#drops duplicated values. First sorts the dataset based on dataset_source(global comes first), then keeps only the first occurrence.
df_combined = (
    df_combined
    .sort_values("dataset_source")
    .drop_duplicates(subset=["canonical_key"], keep="first")
    .reset_index(drop=True)
)

after = len(df_combined)

print(f"Removed {before - after} duplicate songs")
```

```python
with open("df_combined.pkl", "rb") as f:
    df = pickle.load(f)
```

```python
#splitting the dataset back because similarity is computed within a datset
df_global = df[df["dataset_source"] == "global"].reset_index(drop=True)
df_regional = df[df["dataset_source"] == "regional"].reset_index(drop=True)

print(len(df_global), len(df_regional))
```

```python
music_features = [
    "danceability",
    "energy",
    "loudness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "speechiness",
    "tempo"
]

vibe_features = [
    "valence",
    "energy",
    "danceability",
    "tempo"
]
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler_music_global = StandardScaler()
scaler_vibe_global = StandardScaler()

music_matrix_global = scaler_music_global.fit_transform(
    df_global[music_features]
)

vibe_matrix_global = scaler_vibe_global.fit_transform(
    df_global[vibe_features]
)
```

```python
scaler_music_regional = StandardScaler()
scaler_vibe_regional = StandardScaler()

music_matrix_regional = scaler_music_regional.fit_transform(
    df_regional[music_features]
)

vibe_matrix_regional = scaler_vibe_regional.fit_transform(
    df_regional[vibe_features]
)
```

```python
with open("df_global.pkl", "wb") as f:
    pickle.dump(df_global, f)

with open("df_regional.pkl", "wb") as f:
    pickle.dump(df_regional, f)

with open("music_matrix_global.pkl", "wb") as f:
    pickle.dump(music_matrix_global, f)

with open("vibe_matrix_global.pkl", "wb") as f:
    pickle.dump(vibe_matrix_global, f)

with open("music_matrix_regional.pkl", "wb") as f:
    pickle.dump(music_matrix_regional, f)

with open("vibe_matrix_regional.pkl", "wb") as f:
    pickle.dump(vibe_matrix_regional, f)
```
